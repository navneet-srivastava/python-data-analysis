```python
from datetime import datetime , date , time
```

```python
dt = datetime(2011,10,29,20, 30, 59)
print(dt)

dt = datetime(2011,10,29,20, 30, 59, 100000)
print(dt)

```

```python
dt.date()
```

```python
dt.time()
```

```python
print(dt.year)
print(dt.month)
print(dt.day)
print(dt.hour)
print(dt.minute)
print(dt.second)
```

strftime method formats a datetime as string

```python
print(dt.strftime("%Y-%m-%d"))
print(dt.strftime("%Y-%m-%d %H:%M:%S"))
print(dt.strftime("%Y-%m-%d %H:%M:%S %p"))
print(dt.strftime("%Y-%m-%d %H:%M:%S %p %A"))
print(dt.strftime("%Y-%m-%d %H:%M:%S %p %A %B"))
print(dt.strftime("%Y-%m-%d %H:%M:%S %p %A %B %d"))
```

strings can be converted back to datetime objects using strptime method

```python
datetime.strptime('2011-10-29 20:30:59', '%Y-%m-%d %H:%M:%S')
```

When you are aggregating or otherwise grouping time series data, it will occasionally
be useful to replace time fields of a series of datetimes—for example, replacing the
minute and second
fields with zero:



```python
dt_hour = dt.replace(minute=0, second=0,microsecond=0)
dt_hour
```

The difference of two objects produces a type: datetime.timedelta

```python
dt2 = datetime.now()
print(dt2)
print(dt)
delta = dt2 - dt
delta



```
