#### Add ResponseMinutes = arrival minus alarm, in minutes. Flag rows where it is negative or over 120 minutes as suspicious.

```python
df_new = df.copy()
df_new["ResponseMinutes"] = (pd.to_datetime(df_new["arrival_dt_tm"], format="%Y/%m/%d %I:%M:%S %p") - pd.to_datetime(df_new["alarm_dt_tm"], format="%Y/%m/%d %I:%M:%S %p")).dt.total_seconds() / 60
df_new["ResponseMinutes"] = df_new["ResponseMinutes"].fillna(0).astype(int)


df_new.loc[(df_new["ResponseMinutes"] < 0 )|  (df_new["ResponseMinutes"] > 120 ), ["incident_number","ResponseMinutes"]].head(10)

df_new.loc[(df_new["ResponseMinutes"] < 0 ), ["incident_number","ResponseMinutes"]].head(10)

```
