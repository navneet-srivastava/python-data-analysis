```python
import pandas as pd
import numpy as np
```

```python
obj = pd.Series([4, -7, 5, 3])

obj
```

```python
isinstance(obj.dtype,np.int64)
```

isinstance(obj.dtype, np.int64) is False because obj.dtype is not an integer; it is a metadata object describing the data.

Here is the breakdown of what is actually happening under the hood:

np.int64 is a scalar data type class (like Python's built-in int).

obj.dtype is an instance of numpy.dtype. It is a special container object that holds information about the data (byte order, data size, etc.), but it is not the data itself.

If you check the exact type of obj.dtype, you will see it is a dtype object, not an int64

```python
isinstance(obj.dtype,np.dtype)
```

```python
print(type(obj.dtype))
```

If you were to check an actual element inside the Series, isinstance would return True, because the elements themselves are instances of np.int64:

```python
isinstance(obj[0], np.int64)
```

Another way to think about a Series is as a fixed-length, ordered dictionary, as it is a
mapping of index values to data values

```python
sdata = {"Ohio": 35000, "Texas": 71000, "Oregon": 16000, "Utah": 5000}

obj3 = pd.Series(sdata)

obj3
```

```python
"Ohio" in obj3
```

```python
obj3.to_dict()
```

When you are only passing a dictionary, the index in the resulting Series will respect
the order of the keys according to the dictionary’s
method, which depends on
keys
the key insertion order. You can override this by passing an index with the dictionary
keys in the order you want them to appear in the resulting Series:

```python
states = ["California", "Ohio", "Oregon", "Texas"]
obj4 = pd.Series(sdata, index=states)
obj4


```

```python
pd.isna(obj4)
```

```python
pd.notna(obj4)
```

Series also has these as instance methods:

```python
obj4.isna()
```

```python
obj3,obj4
```

```python
obj3 + obj4
```

attribute, which integrates with
